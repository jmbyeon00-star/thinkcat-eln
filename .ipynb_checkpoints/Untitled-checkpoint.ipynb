```python
mapping_path = "/home/yckim/development/ipforce_next/gpu_backend/users/1/models/classification/1/mapping.json"
with open(mapping_path, "r") as f:
    label_map_str = json.load(f)
    label_map = {int(k): v for k, v in label_map_str.items()}
    print(f">>>>> Loaded label_map: {label_map}")
        
epochs = 10
max_length = 256
batch_size = 128
learning_rate = 1e-5
num_labels = len(label_map_str)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

ckpt_tok = "/home/yckim/development/ipforce_next/gpu_backend/models/bert-base-multilingual-trained"
ckpt_model = "/home/yckim/development/ipforce_next/gpu_backend/users/1/models/classification/1/best_model"
# model = BertForSequenceClassification.from_pretrained(ckpt_model, num_labels=num_labels, use_safetensors=True, device_map=None, torch_dtype=torch.float32)
model = BertForSequenceClassification.from_pretrained(ckpt_model, num_labels=num_labels)
tokenizer = tokenizer = BertTokenizerFast.from_pretrained(ckpt_tok, use_fast=True)
model.to(device)
optimizer = AdamW(model.parameters(), lr=learning_rate)
```

```python
import pymysql

user_id = 1
file_id = 585

connection = pymysql.connect(
    host="192.168.1.20",
    user="root",
    password="doslvkdlqm!",
    db="ipforce",
    charset="utf8mb4",
    cursorclass=pymysql.cursors.DictCursor
)
cursor = connection.cursor()

query = """
    SELECT title, abstract, source, target
    FROM FILE_DATA_TB
    WHERE file_id = %s
    ORDER BY id
"""
cursor.execute(query, (file_id,))
result = cursor.fetchall()

cursor.close()
connection.close()
```

```python
df_test = pd.DataFrame(result)

# df_test["source"] = df_test["source"].map(safe_encode_decode)
df_test = df_test.drop_duplicates(subset=["source"], keep="first", ignore_index=True)
df_test = df_test.dropna(subset=["source"]).query("source != ''")
```

```python
model.eval()
model.to(device)

print(">>>>> batch_size", batch_size)
print(">>>>> max_length", max_length)
print(">>>>> tokenizer:", ckpt_tok)
print(">>>>> model:", ckpt_model)

# 데이터 준비
texts = df_test["source"].astype(str).tolist()

# 실제 정답값이 있는지 확인
has_ground_truth = "target" in df_test.columns
if has_ground_truth:
    ground_truths = df_test["target"].astype(str).tolist()
    print(f">>>>> Ground truth detected. Will calculate accuracy.")

results = []
correct_count = 0
total_count = len(texts)

# 배치 단위로 추론 수행
for i in range(0, len(texts), batch_size):
    batch_texts = texts[i:i+batch_size]
    batch_gt = ground_truths[i:i+batch_size] if has_ground_truth else None

    # 토크나이징
    encoded = tokenizer(
        batch_texts, 
        padding=True, 
        truncation=True, 
        max_length=max_length, 
        return_tensors="pt"
    ).to(device)

    # 추론
    with torch.no_grad():
        outputs = model(**encoded)
        probs = torch.softmax(outputs.logits, dim=-1)
        preds = torch.argmax(probs, dim=-1).cpu().tolist() # 상위 3개 클래스
        topk_probs, topk_indices = torch.topk(probs, k=3, dim=-1)  # 상위 3개 클래스
    print("preds:", preds, len(preds))
    print("topk_probs:", topk_probs)
    print("topk_indices:", topk_indices)
    break

    # 결과 저장
    # for idx, (text, pred, prob_dist) in enumerate(zip(batch_texts, preds, probs.cpu().tolist())): # 상위 3개 클래스
    for idx, (text, top_probs, top_indices) in enumerate(zip(batch_texts, topk_probs.cpu().tolist(), topk_indices.cpu().tolist())): # 상위 3개 클래스
        result = {
            "source": text,
            # "label": int(pred), # 상위 3개 클래스
            # "prob": max(prob_dist) # 상위 3개 클래스
            "labels": [int(label) for label in top_indices]
            "probs": [float(p) for p in top_probs], # 상위 3개 클래스 # 각 클래스의 확률
            "top_label": int(top_indices[0]), # 상위 3개 클래스 # 가장 높은 확률의 클래스 (기존 label)
            "top_prob": float(top_probs[0]) # 상위 3개 클래스 # 그 확률
        }

        # 실제 정답이 있으면 추가 정보 저장
        if has_ground_truth:
            gt = batch_gt[idx]
            result["ground_truth"] = gt

            # 정답 비교: 레이블 이름으로 비교 또는 레이블 인덱스로 비교
            predicted_label_name = label_map.get(pred, str(pred))
            result["is_correct"] = (predicted_label_name == gt) or (str(pred) == gt)

            if result["is_correct"]:
                correct_count += 1

        results.append(result)

    # 진행률 업데이트
    progress = min(100, int((i + len(batch_texts)) / total_count * 100))
    print(">>> progress:", progress)

# 평가 지표 계산 (정답이 있는 경우)
evaluation_metrics = None
if has_ground_truth:
    accuracy = (correct_count / total_count) * 100
    evaluation_metrics = {
        "accuracy": accuracy,
        "correct_count": correct_count,
        "total_count": total_count,
        "error_count": total_count - correct_count
    }
    print(f">>>>> Evaluation - Accuracy: {accuracy:.2f}% ({correct_count}/{total_count})")

# label_map을 문자열 키로 변환 (JSON 직렬화를 위해)
mapper_str = {str(k): v for k, v in label_map.items()}

# 결과를 딕셔너리로 구성
output_data = {
    "results": results,
    "evaluation": evaluation_metrics,
    "mapper": mapper_str
}

# 결과를 JSON 파일로 저장
# os.makedirs(result_path, exist_ok=True)
# result_file_path = f"{result_path}/result_classification.json"

# try:
#     with open(result_file_path, "w", encoding="utf-8") as f:
#         json.dump(output_data, f, ensure_ascii=False, indent=2)
#     print(f">>>>> 결과 저장 완료: {result_file_path}")
# except Exception as e:
#     print(f">>>>> 결과 저장 실패: {e}")

# 완료 상태 업데이트
# safe_create_task(self._progress({
#     "progress": 100, 
#     "status": "COMPLETED", 
#     "remaining_time": "0:00:00",
#     # "evaluation": evaluation_metrics  # 평가 지표 추가
# }))
# safe_create_task(self._status({"status": "COMPLETED"}))

# 메모리 정리
gc.collect()
torch.cuda.empty_cache()
```
